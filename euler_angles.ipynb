# Extrinsic Z-Y-Z Euler Angles

This notebook applies extrinsic Z-Y-Z Euler angles to a 3D vector.

For extrinsic rotations, the rotation matrices are applied in the order they are written (left to right), rotating around the **fixed** global axes.

```python
import numpy as np

def rotation_matrix_z(angle):
    """Rotation matrix around the Z-axis."""
    c, s = np.cos(angle), np.sin(angle)
    return np.array([
        [c, -s, 0],
        [s,  c, 0],
        [0,  0, 1]
    ])

def rotation_matrix_y(angle):
    """Rotation matrix around the Y-axis."""
    c, s = np.cos(angle), np.sin(angle)
    return np.array([
        [ c, 0, s],
        [ 0, 1, 0],
        [-s, 0, c]
    ])

def apply_extrinsic_zyz_euler(vector, alpha, beta, gamma):
    """
    Apply extrinsic Z-Y-Z Euler angles to a 3D vector.
    
    For extrinsic rotations, we rotate around fixed global axes.
    The rotation is applied as: R = Rz(gamma) @ Ry(beta) @ Rz(alpha)
    
    Parameters:
    -----------
    vector : array-like
        3D vector to rotate
    alpha : float
        First rotation angle around Z-axis (radians)
    beta : float
        Second rotation angle around Y-axis (radians)
    gamma : float
        Third rotation angle around Z-axis (radians)
    
    Returns:
    --------
    numpy.ndarray
        Rotated 3D vector
    """
    v = np.array(vector)
    
    # Extrinsic: apply rotations in order around fixed axes
    # R_total = Rz(gamma) @ Ry(beta) @ Rz(alpha)
    R = rotation_matrix_z(gamma) @ rotation_matrix_y(beta) @ rotation_matrix_z(alpha)
    
    return R @ v
```

```python
# Example usage
vector = [1, 1, 0.74]  # Unit vector along X-axis
angles = [58.237385, 111.541933, 122.169921]

# Define Euler angles (in radians)
alpha = np.radians(angles[0])   # First Z rotation
beta = np.radians(angles[1])    # Y rotation
gamma = np.radians(angles[2])   # Second Z rotation

# Apply the rotation
rotated_vector = apply_extrinsic_zyz_euler(vector, alpha, beta, gamma)

print(f"Original vector: {vector}")
print(f"Euler angles (Z-Y-Z): α={np.degrees(alpha):.1f}°, β={np.degrees(beta):.1f}°, γ={np.degrees(gamma):.1f}°")
print(f"Rotated vector: [{rotated_vector[0]:.4f}, {rotated_vector[1]:.4f}, {rotated_vector[2]:.4f}]")
```

```python
import numpy as np
from scipy.spatial.transform import Rotation as R

# Your original angles (in degrees)
a, b, g = 30, 45, 60  # example values

# Create ZYZ rotation (scipy uses lowercase 'zyz')
rot_original = R.from_euler('ZYZ', [a, b, g], degrees=True)

# Rotation that takes [-1,0,0] to [0,0,1]: 90° around Y
rot_extra = R.from_euler('Y', 90, degrees=True)

# Compose: pre-multiply
rot_new = rot_extra * rot_original

# Extract new ZYZ angles
a_new, b_new, g_new = rot_new.as_euler('ZYZ', degrees=True)

print(f"New angles: α={a_new:.2f}°, β={b_new:.2f}°, γ={g_new:.2f}°")
```
